```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

T = 1
N = 100
dt = T / N
t = np.linspace(0, T, N)

num_simulations = 1000

sum_df_ito = np.zeros(N-1)
sum_df_taylor = np.zeros(N-1)
sum_df_actual = np.zeros(N-1)

for _ in range(num_simulations):
  dW = np.random.normal(0, np.sqrt(dt), N)
  W = np.insert(np.cumsum(dW[:-1]), 0 ,0)

  f_val = W**2

  df_ito_single_path = 2 * W[:-1] * dW[:-1] + dt
  df_taylor_single_path = 2 * W[:-1] * dW[:-1]
  df_actual_single_path = np.diff(f_val)

  sum_df_ito += df_ito_single_path
  sum_df_taylor += df_taylor_single_path
  sum_df_actual += df_actual_single_path


avg_df_ito = sum_df_ito / num_simulations
avg_df_taylor = sum_df_taylor / num_simulations
avg_df_actual = sum_df_actual / num_simulations

fig, ax = plt.subplots(figsize=(10,6))
ax.plot(t[1:], avg_df_ito, color="cyan", label="Average Ito's Lemma", linestyle="--", linewidth=2, alpha=0.8)
ax.plot(t[1:], avg_df_taylor, color="magenta", label="Average Taylor", linestyle=":", linewidth=2, alpha=0.8)
ax.plot(t[1:], avg_df_actual, color="yellow", label="Actual Average", linewidth=2, alpha=0.8)
ax.plot(t[1:], np.full_like(t[1:], dt), label="Theoretical Expectation", color="#17becf", linewidth=2, alpha=0.6)

ax.set_title(f"Comparison of the Average of Differentials (over {num_simulations} simulations)", fontweight="bold", fontsize=14, pad=16)
ax.set_xlabel("Time")
ax.set_ylabel("Average df")
ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.3)
ax.legend(loc="upper left", fontsize=10, framealpha=0.9)
ax.set_xlim(0, 1.02)

plt.style.use("dark_background")
plt.tight_layout()
plt.show()
```
