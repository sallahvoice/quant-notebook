```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import norm

mu, sigma = 0, 1
N = 512
x_min, x_max = mu - 4 * sigma, mu + 4 * sigma
x = np.linspace(x_min, x_max, N)
dx = x[1] - x[0]

pdf = norm.pdf(x, mu, sigma)

fft_result = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(pdf))) * dx
freq = np.fft.fftshift(np.fft.fftfreq(N, d=dx)) * 2 * np.pi

charc_func_real = np.real(fft_result)
charc_func_imag = np.imag(fft_result)

reconstructed_pdf = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(fft_result))) / dx
reconstructed_pdf = np.real(reconstructed_pdf)


fig2 = make_subplots(
    rows=2, cols=2,
    subplot_titles=[
        "Original PDF (Normal RV)",
        "Characteristic Fuction: Re[ϕ(t)]",
        "Characteristic Fuction: Im[ϕ(t)]",
        "Reconstructed PDF (Inverse FFT)"
        ]
)

fig2.add_trace(go.Scatter(
    x=x, y=pdf,
    name="PDF",
    mode="lines",
    line=dict(color="#70e000", width=3),
), row=1, col=1)

fig2.add_trace(go.Scatter(
    x=freq, y=charc_func_real,
    name="Re[ϕ(t)]",
    mode="lines",
    line=dict(color="#ffd60a", width=3),
), row=1, col=2)

fig2.add_trace(go.Scatter(
    x=freq, y=charc_func_imag,
    name="Im[ϕ(t)]",
    mode="lines",
    line=dict(color="#f15bb5", width=3),
), row=2, col=1)

fig2.add_trace(go.Scatter(
    x=x, y=reconstructed_pdf,
    name="Reconstructed",
    mode="lines",
    line=dict(color="#4cc9f0", width=3),
), row=2, col=2)



fig2.update_layout(
    height=700,
    title="FFT Round-Trip from Normal RV: PDF <-> Characteristic Function <-> Reconstructed PDF",
    showlegend=False,
    plot_bgcolor=("rgba(0,0,0,0)"),
    paper_bgcolor=("rgba(0,0,0,0)"),
    font=dict(color="white"),
)


for r in [1, 2]:
  for c in [1, 2]:
    fig2.update_xaxes(
        showgrid=True,
        gridcolor="rgba(128,128,128,0.3)",
        gridwidth=1,
        zeroline=False,
        row=r,
        col=c
    )

    fig2.update_yaxes(
        showgrid=True,
        gridcolor="rgba(128,128,128,0.3)",
        gridwidth=1,
        zeroline=False,
        row=r,
        col=c
    )

fig2.update_xaxes(title_text="x (domain)", row=1, col=1)
fig2.update_yaxes(title_text="f(x)", row=1, col=1)
fig2.update_xaxes(title_text="t (frequency domain)", row=1, col=2)
fig2.update_yaxes(title_text="Re[ϕ(t)]", row=1, col=2)
fig2.update_xaxes(title_text="t (frequency domain)", row=2, col=1)
fig2.update_yaxes(title_text="Im[ϕ(t)]", row=2, col=1)
fig2.update_xaxes(title_text="x (domain)", row=2, col=2)
fig2.update_yaxes(title_text="f̂(x)", row=2, col=2)

fig2.show()

```
